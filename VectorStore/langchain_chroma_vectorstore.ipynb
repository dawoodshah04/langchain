```python
from sentence_transformers import SentenceTransformer

# Initialize a popular open-source embedding model
# This runs completely locally and does not require an API key or access token
model = SentenceTransformer('all-MiniLM-L6-v2')

# Define the texts you want to embed
texts = [
    "Generating embeddings locally is secure and free.",
    "Open-source models offer great flexibility.",
    "Artificial intelligence is transforming many industries."
]

# Generate the embeddings
embeddings = model.encode(texts)

# Display the shapes of the generated embeddings
for text, embedding in zip(texts, embeddings):
    print(f"Text: '{text}'")
    print(f"Embedding Shape: {embedding.shape}")
    print(f"First 5 dimensions: {embedding[:5]}\n")
```

```python
from langchain_core.documents import Document

# Define the five documents provided
doc1 = Document(
    page_content="Inception is a science fiction thriller directed by Christopher Nolan. The movie follows Dom Cobb, a skilled thief who steals secrets from people's dreams. He is given a chance to erase his criminal history by performing an impossible task called inception.",
    metadata={"genre": "Science Fiction", "director": "Christopher Nolan"}
)

doc2 = Document(
    page_content="Interstellar is a science fiction film directed by Christopher Nolan. It follows a group of astronauts who travel through a wormhole in search of a new habitable planet for humanity. The movie explores themes of love, time, and survival.",
    metadata={"genre": "Science Fiction", "director": "Christopher Nolan"}
)

doc3 = Document(
    page_content="The Shawshank Redemption is a drama film directed by Frank Darabont. It tells the story of Andy Dufresne, a banker who is sentenced to life imprisonment for a crime he did not commit. The film explores friendship, hope, and perseverance.",
    metadata={"genre": "Drama", "director": "Frank Darabont"}
)

doc4 = Document(
    page_content="The Dark Knight is a superhero action film directed by Christopher Nolan. It features Batman fighting against the Joker, a criminal mastermind who creates chaos in Gotham City. The movie explores justice, morality, and sacrifice.",
    metadata={"genre": "Action", "director": "Christopher Nolan"}
)

doc5 = Document(
    page_content="Titanic is a romantic drama film directed by James Cameron. The movie tells the tragic love story of Jack and Rose, two passengers from different social backgrounds who meet aboard the RMS Titanic during its ill-fated maiden voyage.",
    metadata={"genre": "Romance", "director": "James Cameron"}
)

documents = [doc1, doc2, doc3, doc4, doc5]
```

```python
# Extract the page content from each document
contents = [doc.page_content for doc in documents]

# Generate local embeddings for these documents
doc_embeddings = model.encode(contents)

# Display the document metadata alongside their embedding shapes
for i, (doc, emb) in enumerate(zip(documents, doc_embeddings), 1):
    title = doc.page_content.split(' is ')[0]
    print(f"Document {i}: {title}")
    print(f"  Genre: {doc.metadata['genre']}")
    print(f"  Embedding shape: {emb.shape}")
    print(f"  First 3 elements: {emb[:3]}\n")
```

```python
import chromadb

# Initialize the Chroma client
chroma_client = chromadb.PersistentClient(path="./chroma_db")

# Create or get the collection
# Since we already generated embeddings locally, we can insert them directly
collection = chroma_client.get_or_create_collection(name="movie_collection")

# Prepare data for insertion
ids = [f"doc{i}" for i in range(1, len(documents) + 1)]
metadatas = [doc.metadata for doc in documents]
# Convert embeddings array to list for ChromaDB compatibility
embeddings_list = doc_embeddings.tolist()

# Add documents, embeddings, and metadata into the collection
collection.add(
    embeddings=embeddings_list,
    documents=contents,
    metadatas=metadatas,
    ids=ids
)

print(f"Successfully added {collection.count()} documents to ChromaDB!")
```

```python
# Define a search query and embed it using the same open-source model
query = "Are there any space exploration movies directed by Christopher Nolan?"
query_embedding = model.encode([query]).tolist()

# Query the ChromaDB collection
results = collection.query(
    query_embeddings=query_embedding,
    n_results=1
)

# Display the search results
print(f"Query: '{query}'\n")
for doc, metadata, distance in zip(results['documents'][0], results['metadatas'][0], results['distances'][0]):
    print(f"Document: {doc}")
    print(f"Metadata: {metadata}")
    print(f"Distance (Lower is more similar): {distance:.4f}\n")
```

### Implementing FAISS Vector Store
We will install `faiss-cpu` and wrap our local Hugging Face model under LangChain's embedding interface so it integrates directly with LangChain's FAISS class.

```python
!pip install faiss-cpu
```

```python
from langchain_community.vectorstores import FAISS
from langchain_core.embeddings import Embeddings

# Create a wrapper class so LangChain can use our local SentenceTransformer model
class LocalEmbeddings(Embeddings):
    def __init__(self, model):
        self.model = model

    def embed_documents(self, texts):
        return self.model.encode(texts).tolist()

    def embed_query(self, text):
        return self.model.encode([text])[0].tolist()

# Wrap our loaded local model
local_embeddings_wrapper = LocalEmbeddings(model)

# Build the FAISS vector store directly from our LangChain Document objects
db = FAISS.from_documents(documents, local_embeddings_wrapper)

print("FAISS vector store successfully created and populated!")
```

```python
# Define the query
query = "Are there any space exploration movies ?"

# Perform a similarity search with scores (L2 distance)
search_results = db.similarity_search_with_score(query, k=2)

print(f"Query: '{query}'\n")
for doc, score in search_results:
    print(f"Document: {doc}\n")
    print(f"Distance Score (L2 distance): {score:.4f}\n")
```
